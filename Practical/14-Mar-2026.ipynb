```python
"""
Program No.: 7
Experiment 19: A timestamp is three numbers: a number of hours, minutes and seconds. Given two timestamps, calculate how many seconds are between them. The moment of the first timestamp occurred before the moment of the second timestamp.
Author: Siddhesh Chavan
Date: 26/02/2026
"""

print("Timestamp 1:")
hour_1 = int(input("Enter the hour: "))
minutes_1 = int(input("Enter the minutes: "))
seconds_1 = int(input("Enter the seconds: "))

print("\nTimestamp 2:")
hour_2 = int(input("Enter the hour: "))
minutes_2 = int(input("Enter the minutes: "))
seconds_2 = int(input("Enter the seconds: "))

total_seconds_1 = hour_1 * 3600 + minutes_1 * 60 + seconds_1
total_seconds_2 = hour_2 * 3600 + minutes_2 * 60 + seconds_2

difference = total_seconds_2 - total_seconds_1

print(f"\nDifference between two timestamps: {difference} seconds")
```

```python
"""
Program No.: 8
Experiment 14: Program to check if a date is valid or not
Author: Siddhesh Chavan
Date: 26/02/2026
"""

day = int(input("Enter the day: "))
month = int(input("Enter the month: "))
year = int(input("Enter the year: "))

is_leap = (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)
valid = True

# Month Range:
if month < 1 or month > 12:
    valid = False

# Day Range:
elif month in [1, 3, 5, 7, 8, 10, 12]:
    if day < 1 or day > 31:
        valid = False

elif month in [4, 6, 9, 11]:
    if day < 1 or day > 30:
        valid = False

elif month == 2:
    if is_leap: # Only allow 29 days in the month of Feb if it is a leap
        if day < 1 or day > 29:
            valid = False
    else:
        if day < 1 or day > 28:
            valid = False

if valid:
    print(f"The entered date: {day}/{month}/{year} is valid!")
else:
    print("The entered date is invalid!")
```

```python
"""
Program No.: 30
Experiment 59: Function to find the factorial of a number using recursion
Author: Siddhesh Chavan
Date: 14/03/2026
"""

def factorial(num):
    if num <= 1:
        return 1
    return num * factorial(num - 1)

a = int(input("Enter a number: "))
if a < 0:
    print("Factorial of negative numbers doesn't exist!")
else:
    print(f"Factorial of {a} = {factorial(a)}")
```

```python
"""
Program No.: 31
Experiment 60: Function to find the sum of the digits of the number recursively
Author: Siddhesh Chavan
Date: 14/03/2026
"""

def sum_digits(num):
    result = 0
    while num != 0:
        digit = num % 10
        result += digit
        num //= 10
    return result

a = int(input("Enter a number: "))
print(f"Sum of digits of {a} = {sum_digits(a)}")
```

```python
"""
Program No.: 32
Experiment 61: Create a function named 'count_vowels' that accepts a string and returns the number of vowels in the string
Author: Siddhesh Chavan
Date: 14/03/2026
"""

def count_vowels(input_str):
    VOWELS = ["a", "e", "i", "o", "u"]
    count = 0
    for each_char in input_str.lower():
        if each_char in VOWELS:
            count += 1
    return count

str1 = "Python for Engineers"
print(f"Number of vowels in string = {count_vowels(str1)}")
```
