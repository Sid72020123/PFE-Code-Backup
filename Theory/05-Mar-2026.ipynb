```python
# Program to check whether an entered number is positive, negative or zero

num = int(input("Enter a number: "))

if num > 0:
    print("The number is positive")
elif num < 0:
    print("The number is negative")
else:
    print("The number is zero")

# Short Hand if-else conversion:
(print("The number is positive") if num > 0 else print("The number is negative")) if num != 0 else print("The number is zero")
```

```python
# Find out the average of marks of a student in 5 subjects and print out the status like 'pass', 'fail', 'distinction', etc.

marks_1 = int(input("Enter marks of Subject 1: "))
marks_2 = int(input("Enter marks of Subject 2: "))
marks_3 = int(input("Enter marks of Subject 3: "))
marks_4 = int(input("Enter marks of Subject 4: "))
marks_5 = int(input("Enter marks of Subject 5: "))

total_marks = marks_1 + marks_2 + marks_3 + marks_4 + marks_5
average = total_marks / 5

print("Total Marks:", total_marks)
print("Average:", average)
print("Result Status: ", end="") # keep the cursor on the same line

if average >= 80:
    print("Distinction")
elif average >= 60:
    print("First Class")
elif average >= 50:
    print("Second Class")
elif average >= 40:
    print("Pass")
else:
    print("Fail")
```

```python
# for loop: Display table of any given number

num = int(input("Enter a number: "))

for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
```

```python
# for loop: Summation of numbers from a given range

start = int(input("Enter the starting number: "))
end = int(input("Enter the ending number: "))

result = 0
for i in range(start, end + 1):
    result += i

print(f"Sum of all numbers from {start} to {end} = {result}")
```

```python
# for loop: Display all even and odd numbers from a given series and their summation

start = int(input("Enter the starting number: "))
end = int(input("Enter the ending number: "))

odd = []
even = []
for i in range(start, end + 1):
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)

print(f"\nRange [{start} - {end}]:")
print(f"Odd Numbers in range: {', '.join([str(s) for s in odd])}")
print(f"Summation of Odd numbers = {sum(odd)}\n")
print(f"Even Numbers in range: {', '.join([str(s) for s in even])}")
print(f"Summation of Even numbers = {sum(even)}")
```

```python
# for loop: Seperate out the positive and negative numbers from a list

numbers = input("Enter any numbers seperated by space: ").strip().split(" ")
numbers = list(map(lambda x: int(x), numbers))

positive = []
negative = []
summation_positive = 0
summation_negative = 0

for n in numbers:
    if n > 0:
        positive.append(str(n))
        summation_positive += n
    else:
        negative.append(str(n))
        summation_negative += n

print(f"All positive numbers: {', '.join(positive)}")
print(f"Sum of all positive numbers = {summation_positive}\n")
print(f"All negative numbers: {', '.join(negative)}")
print(f"Sum of all negative numbers = {summation_negative}")
```

```python
# while loop: Display table of any given number

num = int(input("Enter a number: "))

i = 1
while i <= 10:
    print(f"{num} x {i} = {num * i}")
    i += 1
```

```python
# while loop: Summation of numbers from a given range

start = int(input("Enter the starting number: "))
end = int(input("Enter the ending number: "))

result = 0
i = start
while i <= end:
    result += i
    i += 1

print(f"Sum of all numbers from {start} to {end} = {result}")
```

```python
# while loop: Display all even and odd numbers from a given series and their summation

start = int(input("Enter the starting number: "))
end = int(input("Enter the ending number: "))

odd = []
even = []
i = start
while i <= end:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)
    i += 1

print(f"\nRange [{start} - {end}]:")
print(f"Odd Numbers in range: {', '.join([str(s) for s in odd])}")
print(f"Summation of Odd numbers = {sum(odd)}\n")
print(f"Even Numbers in range: {', '.join([str(s) for s in even])}")
print(f"Summation of Even numbers = {sum(even)}")
```

```python
# while loop: Seperate out the positive and negative numbers from a list

numbers = input("Enter any numbers seperated by space: ").strip().split(" ")
numbers = list(map(lambda x: int(x), numbers))

positive = []
negative = []
summation_positive = 0
summation_negative = 0

i = 0
while i < len(numbers):
    n = numbers[i]
    if n > 0:
        positive.append(str(n))
        summation_positive += n
    else:
        negative.append(str(n))
        summation_negative += n
    i += 1

print(f"All positive numbers: {', '.join(positive)}")
print(f"Sum of all positive numbers = {summation_positive}\n")
print(f"All negative numbers: {', '.join(negative)}")
print(f"Sum of all negative numbers = {summation_negative}")
```
